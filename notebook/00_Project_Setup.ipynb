### required CSV templates

```python
templates = {
    "grid_substations_template.csv": [
        "bus_id", "name", "latitude", "longitude",
        "voltage_kv"
    ],

    "transmission_lines_template.csv": [
        "line_id", "from_bus", "to_bus",
        "length_km", "voltage_kv",
        "capacity_mw", "resistance_ohm",
        "reactance_ohm"
    ],

    "solar_plants_template.csv": [
        "plant_id", "name", "bus_id",
        "latitude", "longitude",
        "capacity_mw", "annual_generation_gwh"
    ],

    "hourly_demand_template.csv": [
        "timestamp", "bus_id", "demand_mw"
    ],

    "solar_generation_template.csv": [
        "timestamp", "plant_id", "solar_mw"
    ],

    "battery_parameters_template.csv": [
        "parameter", "value", "unit", "source"
    ],

    "system_hourly_profile_template.csv": [
        "timestamp", "demand_mw", "solar_mw"
    ]
}

for filename, columns in templates.items():

    path = ROOT / "datasets" / "templates" / filename

    if not path.exists():
        pd.DataFrame(columns=columns).to_csv(
            path, index=False
        )

    print("Created:", filename)
```

### Check official datasets

```python
official_folder = ROOT / "datasets" / "raw" / "official_report_csv"
if not official_folder.exists() or not any(official_folder.glob("*.csv")):
    raise FileNotFoundError(f"Official CSV directory not found: {official_folder}")

for file in sorted(official_folder.glob("*.csv")):

    df = pd.read_csv(file)

    print("\nDataset:", file.name)
    print("Rows:", len(df))
    print("Columns:", list(df.columns))
```
